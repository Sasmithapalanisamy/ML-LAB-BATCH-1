```python
import numpy as np
from collections import Counter
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
import os
import joblib
import json

SEED = 42
np.random.seed(SEED)

iris = load_iris()
X = iris.data
y = iris.target
target_names = iris.target_names

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=SEED, stratify=y
)

os.makedirs('knn_models', exist_ok=True)

def euclidean_distance(a, b):
    return np.sqrt(np.sum((a - b) ** 2))

def knn_predict(X_train, y_train, x_test_point, k):
    distances = [euclidean_distance(x_test_point, x_train_point) for x_train_point in X_train]
    k_indices = np.argsort(distances)[:k]
    k_nearest_labels = [y_train[i] for i in k_indices]
    most_common = Counter(k_nearest_labels).most_common(1)
    return most_common[0][0]

results = []

for k in [1, 3, 5, 7, 9]:
    print(f"\nTraining Manual KNN with K={k}")
    y_pred= [knn_predict(X_train, y_train, x, k=k) for x in X_test]
    acc = accuracy_score(y_test, y_pred)
    print(f"Manual KNN Accuracy for K={k}: {acc:.4f}")
    print(classification_report(y_test, y_pred, target_names=target_names))
    
    manual_model = {
        'X_train': X_train,
        'y_train': y_train,
        'k': k
    }
    model_path = f'knn_models/manual_knn_k{k}.joblib'
    joblib.dump(manual_model, model_path)
 
    results.append({
        'k': k,
        'accuracy': acc,
        'model_path': model_path,
    })

with open('knn_models/experiment_summary.json', 'w') as f:
    json.dump(results, f, indent=4)

print("\nModels and experiment summary saved in 'knn_models/' directory.")
```
