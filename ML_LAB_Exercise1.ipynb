```
arr=np.array([4,6,8,2,1])
print("Size:",arr.size)
print("Shape:",arr.shape)
print("dimension:",arr.ndim)
print("datatype:",arr.dtype)
print("transpose:",arr.T)
print("ItemSize:",arr.itemsize)
```

numpy Methods


```
print("Sort:",np.sort(arr))
print("SUM:",np.sum(arr))
print("Mean:",np.mean(arr))
print("Maximum",np.max(arr))
print("Minimum:",np.min(arr))
print("Standard Deviation:",np.std(arr))
deg=np.array([0,30,45,60,90])
rad=np.radians(deg)
print("sin:",np.sin(rad))
print("cos:",np.cos(rad))
print("tan:",np.tan(rad))
```

```
val=np.array([-1,0,0.5,1])
print("arcsin:",np.arcsin(val))
print("arccos:",np.arccos(val))
print("arctan:",np.arctan(val))
```

```
mat=np.array([[1,1,3],[0,1,1],[0,0,1]])
print("determinant:",linalg.det(mat))
eig_v,eig_vec=linalg.eig(mat)
print("eigen values:",eig_v)
print("eigen vectors:",eig_vec)
print("Inverse:",linalg.inv(mat))
```
